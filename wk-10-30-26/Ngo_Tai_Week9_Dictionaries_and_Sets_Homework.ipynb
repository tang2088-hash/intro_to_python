**A2. Product Lookup (6 pts)**

Write a function `lookup_price(catalog, product)` that returns the price of `product` if it is a key in `catalog`. Otherwise, it returns the string `"<product> not found"`. Use the `in` operator. Don't use `get()` here.

````python
'''
Type Pseudocode here

CHECK IF PRODUCT IS IN CATALOG
IF IT IS PRESENT, RETURN THE PRICE
OTHERWISE, RETURN MESSAGE "<PRODUCT> NOT FOUND"
'''

def lookup_price(catalog, product):
   if product in catalog:
       return catalog[product]
   else:
       return f"{product} not found"

catalog = {"notebook": 3.99, "pen": 1.25, "backpack": 39.99}
print(lookup_price(catalog, "pen"))
print(lookup_price(catalog, "stapler"))

# Tests
assert lookup_price(catalog, "pen") == 1.25
assert lookup_price(catalog, "stapler") == "stapler not found"
print("✅ A2 passed")
````

**B2. Receipt (6 pts)**

Write a function `order_total(prices, order)` that takes a dictionary of prices and a **list** of items ordered, and returns the total, rounded to 2 decimals. Items that aren't in `prices` should be **skipped**. Use a `for` loop.

````python
'''
Type Pseudocode here

SET TOTAL TO 0
CREATE LOOP THROUGH ORDERED ITEMS
ADD PRICE WHEN ITEMS EXISTS IN PRICES
RETURN ORDER TOTAL ROUNDED TO 2 DECIMALS
'''
def order_total(prices, order):
    total = 0

    for item in order:
        if item in prices:
            total += prices[item]

    return round(total, 2)
    
menu = {"Latte": 4.50, "Mocha": 4.75, "Cold Brew": 5.00, "Matcha": 5.25}
print(order_total(menu, ["Latte", "Latte", "Matcha"]))

# Tests
assert order_total(menu, ["Latte", "Latte", "Matcha"]) == 14.25
assert order_total(menu, ["Mocha", "Pizza"]) == 4.75
assert order_total(menu, []) == 0
print("✅ B2 passed")
````

**B4. Word Counter (6 pts)**

Write a function `count_words(text)` that returns a dictionary mapping each word in `text` to the number of times it appears. Make the words lowercase first and split on spaces with `text.lower().split()`.

````python
'''
Type Pseudocode here

CREATE EMPTY DICTIONARY
CREATE LOOP TO MAKE TEXT LOWERCASE AND SPLIT ON SPACES
INCREASE COUNT FOR EACH WORD
RETURN DICTIONARY
'''
def count_words(text):
    counts = {}

    for word in text.lower().split():
        counts[word] = counts.get(word, 0) + 1

    return counts
    
review = "Great coffee great service and great prices"
print(count_words(review))

# Tests
assert count_words(review) == {"great": 3, "coffee": 1, "service": 1, "and": 1, "prices": 1}
assert count_words("") == {}
print("✅ B4 passed")
````

**B5. Top Seller (4 pts)**

Write `top_seller(sales)` that returns a **tuple** `(product, amount)` for the product with the highest sales. Use a loop. Don't use `max()`.

````python
'''
Type Pseudocode here

DEFINE TOP SELLER WITH SALES
SET BEST PRODUCT TO EMPTY STRING
SET HIGHEST AMOUNT TO NEGATIVE INFINITY
CREATE LOOP THROUGH EAHC PRODUCT AND SALES AMOUNT
IF SALES AMOUNT IS GREATER THAN HIGHEST SALES AMOUNT
SAVE BEST PRODUCT HAS TOP SELLER
SAVE SALES AMOUNT AS NEW HIGHEST
RETURN TOP SELLER PRODUCT AND SALES AMOUNT AS TUPLE
'''
def top_seller(sales):
    best_product = ""
    highest_amount = float("-inf")
    
    for product, amount in sales.items():
        if amount > highest_amount:
            best_product = product
            highest_amount = amount

    return (best_product, highest_amount)

sales = {"Latte": 320, "Mocha": 210, "Cold Brew": 405, "Matcha": 150}
print(top_seller(sales))

# Tests
assert top_seller(sales) == ("Cold Brew", 405)
assert top_seller({"Tea": 5}) == ("Tea", 5)
print("✅ B5 passed")
````

**C2. Comprehensions (10 pts)**

Use a **dictionary comprehension** for each part. Each answer should be a single line.

````python
'''
Type Pseudocode here

CREATE DICTIONARY WITH resort AND price WITH 10% DISCOUNT
ROUND DISCOUNTED PRICE TO 2 DECIMALS

CREATE RESORTS DICTIONARY WITH ORIGINAL PRICES UNDER 175

CREATE DICTIONARY WITH EACH product NAME AS KEY
USE NUMBER OF CHARACTERS IN NAME AS VALUE
'''
prices = {"Eldora": 129, "Copper Mountain": 159, "Breckenridge": 179, "Vail": 229}
products = ["notebook", "pen", "backpack"]

# C2a (3 pts): every lift ticket with a 10% discount, rounded to 2 decimals
discounted = {resort: round(price*0.90, 2) for resort, price in prices.items()}

# C2b (3 pts): only resorts that cost under $175
budget = {resort: price for resort, price in prices.items() if price < 175}

# C2c (4 pts): map each product name to its length (keys come from the list)
name_lengths = {product: len(product) for product in products}

print(discounted)
print(budget)
print(name_lengths)

# Tests
assert discounted == {"Eldora": 116.1, "Copper Mountain": 143.1, "Breckenridge": 161.1, "Vail": 206.1}
assert budget == {"Eldora": 129, "Copper Mountain": 159}
assert name_lengths == {"notebook": 8, "pen": 3, "backpack": 8}
print("✅ C2 passed")
````

**D1. Unique Customers (6 pts)**

`visits` lists the customer ID for every visit last week. Write `unique_customers(visits)` that returns a **sorted list** of the unique IDs. Use a set.

````python
'''
Type Pseudocode here

INPUT DEFINE unique_customers WITH visits
CONVERT visits TO A SET TO REMOVE DUPLICATES
SORT SET AND RETURN AS LIST
'''
def unique_customers(visits):
    return sorted(set(visits))


visits = [104, 230, 104, 312, 230, 104, 451, 312]
print(unique_customers(visits))

# Tests
assert unique_customers(visits) == [104, 230, 312, 451]
assert unique_customers([]) == []
print("✅ D1 passed")
````

**D3. Store Comparison (10 pts)**

Two stores carry these products. Use set operators **or** methods to build each set.

````python
'''
Type Pseudocode here

FIND PRODUCTS IN both_stores
FIND PRODUCTS IN any_store
FIND PRODUCTS IN boulder_only
FIND PRODUCTS IN one_store
'''
boulder = {"laptop", "mouse", "keyboard", "webcam", "headset"}
denver  = {"keyboard", "monitor", "mouse", "printer"}

both_stores  = boulder & denver   # in both stores
any_store    = boulder | denver   # in at least one store
boulder_only = boulder - denver   # only in Boulder
one_store    = boulder ^ denver   # in exactly one of the stores

print(both_stores, any_store, boulder_only, one_store, sep="\n")

# Tests
assert both_stores == {"mouse", "keyboard"}
assert any_store == {"laptop", "mouse", "keyboard", "webcam", "headset", "monitor", "printer"}
assert boulder_only == {"laptop", "webcam", "headset"}
assert one_store == {"laptop", "webcam", "headset", "monitor", "printer"}
print("✅ D3 passed")
````

**D4. Job Requirements (3 pts)**

Write `is_qualified(candidate_skills, required_skills)` that returns `True` if the candidate has **every** required skill. Use `issubset` / `issuperset` or `<=` / `>=`.

````python
'''
Type Pseudocode here

INPUT DEFINE is_qualified WITH candidate_skills AND required_skills
CHECK IF ALL required_skills ARE IN SET candidate_skills
RETURN True OR False
'''
def is_qualified(candidate_skills, required_skills):
    return required_skills.issubset(candidate_skills)


required = {"python", "sql"}
print(is_qualified({"python", "sql", "excel"}, required))
print(is_qualified({"python", "tableau"}, required))

# Tests
assert is_qualified({"python", "sql", "excel"}, required) == True
assert is_qualified({"python", "tableau"}, required) == False
print("✅ D4 passed")
````

## Part F: Pickling (10 pts)

Write two functions:
- `save_data(filename, obj)` pickles `obj` into `filename`.
- `load_data(filename)` unpickles and **returns** the object in `filename`.

Remember to open the file in **binary** mode.

````python
'''
Type Pseudocode here

IMPORT PICKLE

INPUT DEFINE save_data WITH filename AND object
OPEN FILE IN BINARY WRITE MODE (wb)
PICKLE object INTO FILE

INPUT DEFINE load_data WITH filename
OPEN FILE IN BINARY READ MODE (rb)
UNPICKLE object AND RETURN
'''
import pickle


def save_data(filename, obj):
    with open(filename, "wb") as file:
        pickle.dump(obj, file)

def load_data(filename):
    with open(filename, "rb") as file:
        return pickle.load(file)

contacts = {"Chris": "555-1111", "Katie": "555-2222", "Joanne": "555-3333"}
save_data("contacts.dat", contacts)
loaded = load_data("contacts.dat")
print(loaded)

# Tests
assert loaded == contacts
assert type(loaded) == dict
save_data("tags.dat", {"vip", "new"})
assert load_data("tags.dat") == {"vip", "new"}
print("✅ Part F passed")
````
